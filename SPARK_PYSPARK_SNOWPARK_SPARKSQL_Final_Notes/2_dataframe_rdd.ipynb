- RDDs are the main logical data units in Spark. They are a distributed collection of objects, which are stored in memory or on disks of different machines of a cluster.

**Practice Differences between RDD Vs DataFrame**


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
#Initialize Spark Session
spark = SparkSession.builder.appName("RDD Vs DF").getOrCreate()
```

```python
spark
```

```python
#Sample data in list
data = [(1, "Alice", 20), (2, "Bob", 30), (3, "Charlie", 10), (4, "David", 40)]
```

```python
# Create RDD
RDD=spark.sparkContext.parallelize(data)
#RDD.collect()
rdd_filtered=RDD.filter(lambda x: x[2] >20)
rdd_result=rdd_filtered.collect()

for row in rdd_result:
    print(row)
```

```python
column=["id","name","age"]
df=spark.createDataFrame(data,column)
df_filtered=df.filter("age<30")
#df.show()
df_filtered.show()
```

**#Difference 1:**
    #1. RDD Creation: We use sparkContext.parallelize(data) to create an RDD. This gives us low-level control over the dataset and operations.
    #1. DataFrame Creation: We use spark.createDataFrame(data, columns) to create a DataFrame, which provides higher-level abstraction and optimizations.

#Difference 2:
    #we collect the RDD result using .collect() and show the DataFrame result using .show().

#Note:
    #API: RDDs use functional programming constructs, while DataFrames are closer to SQL operations and provide higher-level abstractions.

#When working with structured data, DataFrames are usually preferred for performance reasons, while RDDs are still useful for unstructured data or low-level transformations.

**Dataframe vs Dataset:**
- Dataframe are SQL api of Spark. It uses Kryo serializer at the backend. It uses  Tungsten for optimization.
- Dataset is kind of combination of RDD and Dataframe. It uses Encoders for optimization.
