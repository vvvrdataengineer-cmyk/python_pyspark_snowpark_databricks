```python
#13
# Parquet, ORC&AVRO.
# Understand Row and Columnar Data storage formats.
# Read Parquet and ORC data files.
# Benefits of Parquet columnar data format.

# Q. Difference between row and columnar storage?

#Parquet:
    # compression is better.
    # storage is columnar.
    # schema evolution is good.
    # use for DELTA LAKE/SPARK.

#AVRO:
    # compression is good.
    # storage is row.
    # schema evolution is best.
    # used for KAFKA.

df_parquet=spark.read.format("parquet")
					.load("data/input/emp.parquet")
df_parquet1=spark.read.format("parquet")
					.load("data/input/")   # to read all files from folder.
df_parquet2=spark.read.format("parquet")
					.load("data/input/*.parquet")   # to read only parquet files from folder.
     
#df_parquet.show()
df_orc=spark.read.format("orc")
					.load("data/input/emp.orc")
#df_orc.show()

# Program:
import time
def get_time(func):
    def inner_get_time() -> str;
        start_time = time.time()
        func()
        end_time = time.time()
        return (f"Execution time: {(end_time - start_time)*1000} ms")
    print(inner_get_time)

@get_time
def x():
    df = spark.read.format("parquet").load("data/source/emp.parquet")
    df.count()

@get_time
def x():
    df = spark.read.format("parquet").load("data/source/emp.parquet")
    df.select("emp_id").count()  # will get benifit if read specific columns in select from columnar storage.

#recursive read of parquet files.

df_parquet_resursive=spark.read.format("parquet")
                    .option("recursiveFieldLookup",True)
					.load("data/input/recursive_folder/")
```

```python
#16 

# Create the kafka df to read from kafka.
kafka_df = (
	spark
	.readStream
	.format("kafka")
	.option("kafka.bootstrap.servers","ed-kafka:29092")
	.option("subscribe","wildlife")
	.option("startingoffsets","earliest")
	.load()
)

# convert binary to string value column.
kafka_json_df =kafka_df.withColumn("value",expr("cast(value as string)"))

from pyspark.sql.functions import from_json, col, split, explode 

#json schema 
json_schema = "event_time string, data string"

# Expand JSON from value column using schema.
json_df = kafka_json_df.withColumn("values_json", from_json(col("value"), json_schema))

#select the required columns 
flattened_df = json_df.select("values_json.event_time","values_json.data")

#split the data in words.
	words_df = flattened_df \
		.withColumn("words",split("data"," ") \
		.withColumn("word",explode("words")) \
		.withColumn("event_time",col("event_time").cast("timestamp"))
		
from pyspark.sql.functions import count,lit, window 
df_agg = words_df \
	.withwatermark("event_time", '10 minutes") \                                     
	.groupBy(window("event_time", '10 minutes"),"word").agg(count(lit(1)).alias("cnt"))
	
df_final = df_agg.selectExpr("window.start as start_time","window.end as end_time","word","cnt")
df_final.printSchema()

(
df_final
.writeStream
.format("console")
.outputMode("complete")
.trigger(processingTime='30 seconds')
.option("checkpointLocation","checkpoint_dir_kafka_2")
.start()
)
# complete mode does not work with watermark.

(
df_final
.writeStream
.format("console")
.outputMode("update")
.trigger(processingTime='30 seconds')
.option("checkpointLocation","checkpoint_dir_kafka_3")
.start()
)


#Sliding window
df_agg = words_df \
	.withwatermark("event_time", '10 minutes") \                                     
	.groupBy(window("event_time", '10 minutes","15 minutes"),"word").agg(count(lit(1)).alias("cnt"))
	
#"15 minutes" is sliding window.

```
