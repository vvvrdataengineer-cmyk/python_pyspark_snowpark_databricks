**Q. what is Pyspark?**
- PySpark is an interface for Apache Spark in Python.
- Pyspark Supports Spark SQL, DataFrame, Streaming, MLlib (Machine Learning) and Spark Core.



**Spark Commands:**
- pyspark
- spark-shell
- spark-submit
- spark-sql


```python



data = [(1111,'akshara'),(2222,'sadhana')]
df = spark.createDataFrame(data=data)
df.show()

schema = ['id','ename']
df = spark.createDataFrame(data,schema)
df.show()

from pyspark.sql.types import *
data = [{'empid':1111,'ename':'akshara'},{'empid':2222,'ename':'akshara'}]
df = spark.createDataFrame(data)  # Data Types will be Infer automatically based on data.
df.show()
df.printSchema()

from pyspark.sql.types import *
data = [{'empid':1111,'ename':'akshara'},{'empid':2222,'ename':'akshara'}]
schema = StructType([StructField(name='empid',dataType=IntegerType()),
                    StructField(name='ename',dataType=StringType())
                    ])
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```
