**1. What is Snowpark?**

**Snowpark is a developer API provided by Snowflake that allows you to process data using programming languages like:**

- Python
- Scala
- Java

- Unlike traditional SQL-only workflows, Snowpark enables data transformations using code while executing inside Snowflake.

**Key Point:**

- Pandas → Library (local processing)
- Apache Spark → Distributed computing framework
- Snowpark → API that runs inside Snowflake (pushdown execution)

**Pandas vs PySpark vs Snowpark**

```python
# python 
def model(dbt, session):
    dbt.config(materialized="table")
    df = dbt.ref("emp")
    return df
```

```python
# Using UDF in Snowpark
from snowflake.snowpark.functions import udf

def model(dbt, session):
    dbt.config(materialized="table")

    @udf
    def add_one(x: int) -> int:
        x = 0 if not x else x
        return x + 1

    df = dbt.ref("emp")
    df = df.withColumn("newcolumn_name", add_one(df["empid"]))
    return df
```

**Module 4: Snowpark Python Worksheets:**

**1. What are Python Worksheets?**

**Python Worksheets in Snowflake (Snowsight) allow you to:**

- Write Snowpark code directly inside Snowflake
- Develop without local setup
- Build pipelines end-to-end inside Snowflake

**Code runs inside Snowflake, not locally**

**Python Worksheet → Snowpark API → SQL → Snowflake Execution**

**2. Basic Python Worksheet Structure:**

- main(session) is mandatory handler
- Return value appears in Results tab
- Full script runs — no partial execution

```python
import snowflake.snowpark as snowpark
from snowflake.snowpark.functions import col

def main(session: snowpark.Session):
    df = session.table("information_schema.packages")\
        .filter(col("language") == "python")

    df.show()
    return df
```

```python
# Program 1: Create Table from Range

def main(session):
    df = session.range(1, 10, 2).to_df("a")
    df.write.mode("overwrite").save_as_table("range_table")
    return "Table created"
```

```python
# Program 2: Aggregation Pipeline

def main(session):
    df = session.table("task_history")

    df.filter(col("STATE") != "SKIPPED")\
      .group_by("SCHEDULED_TIME", "STATE")\
      .count()\
      .write.mode("overwrite")\
      .save_as_table("aggregate_task_history")

    return "Aggregation complete"
```
