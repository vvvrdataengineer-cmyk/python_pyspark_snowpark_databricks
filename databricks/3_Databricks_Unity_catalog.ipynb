**Q. what is databricks unity catalog?**

- Unity catalog is a unified governance solution for data and AI assest on databricks.
- Open source.
- Unity catalog provides centralied access control, auditing, lineage and discovery capabilities across databricks workspaces.
- fine grained access control (who can see / use what).
- Unity catalog to maintain metadata.
- single metastore across multiple workspaces.
- catalog is part of unity catalog can create based on need. like for DEV, QA, UAT, PROD.

**With Unity Catalog:**
- Central governance.
- Secure sharing across teams.
- Easier compliance (Audit, PII Access).
- Clean separation by domain or environment.

**Without Unity Catalog:**
- Seperate Hive metastore per workspace.
- Duplicate tables.
- Manual permission handling.
- No centralized audit or Lineage.

**Managed Tables:**
- Created under the database directly.
- Dropping the table, delete the underlying data files.

- create table t1();

**External Tables:**
- Created outside the database directory.
- Dropping the table, will not delete the underlying data files.

- create table t1() location 'path of files';

**Note:**
- But metadata available in metastore.
