```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_bookings = spark.read.format("delta")\
  .load("/Volumes/workspace/bronze/bronze_volume/bookings/data/")

#display(df_bookings)
#display(df_bookings.schema)
#print(df_bookings.schema)
```

```python
df=df_bookings.withColumn("amount",col("amount").cast(DoubleType()))\
    .withColumn("modified_date",current_timestamp())\
    .withColumn("booking_date",to_date(col("booking_date")))\
    .drop("_rescued_data")
display(df)
display(df.schema)
```
